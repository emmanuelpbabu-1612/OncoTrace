# OncoTrace: METABRIC Data Audit

Inspect the dataset before building survival models.


```python
from pathlib import Path
import pandas as pd

data_path = Path("../data/raw/METABRIC_RNA_Mutation.csv")
if not data_path.exists():
    raise FileNotFoundError(data_path.resolve())

df = pd.read_csv(data_path, low_memory=False)
print("Shape:", df.shape)
print("\nOverall survival coding:")
print(df["overall_survival"].value_counts(dropna=False).sort_index())
print("\nDeath-from-cancer values:")
print(df["death_from_cancer"].value_counts(dropna=False))
print("\nSurvival months:")
print(df["overall_survival_months"].describe())
```
